## All The Null Values re Fixed

##1. Now fixing and cleaning the Lyrics Column

```python
import re
def clean_lyrics(text):
    text=str(text).lower()
    text=re.sub(r'\[.*?\]','',text)
    text=re.sub(r'\n','',text)
    text=re.sub(r'[^a-z\s]','',text)
    text=re.sub(r'\s+',' ',text)
    return text.strip()
```

```python
df['lyrics_clean']=df['lyrics'].apply(clean_lyrics)
```

#### Feature engineering Part

```python
conditions=[
    (df['valence'] >= 0.5) & (df['energy'] >= 0.5),
    (df['valence'] < 0.5) & (df['energy'] >= 0.5),
    (df['valence'] >= 0.5) & (df['energy'] < 0.5),
    (df['valence'] < 0.5) & (df['energy'] < 0.5)
]
choices=['Happy','Angry','Calm','Sad']
df['mood']=np.select(conditions,choices,default='Sad')
```

```python
sns.scatterplot(df['tempo'])
plt.show()
```

```python
tempo_condition=[
    df['tempo'] < 90,
    (df['tempo'] >= 90) & (df['tempo'] <= 130),
    df['tempo'] >= 130 
]
```

```python
tempo_choices=['slow','medium','fast']
df['tempo_cat']=np.select(tempo_condition,tempo_choices,default='medium')
```

```python
df['lyrics_length']=df['lyrics_clean'].apply(lambda x:x.split())
```

## Some Audion features Have different values of Maximum and MiniMum value so we are using scaling --> MinMaxSacler

## Bringing all the Features ot a range of (0,1)

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
fig,axes=plt.subplots(1,2,figsize=(20,7))
df[features].boxplot(ax=axes[0])
axes[0].set_title('Before Min-Max-Scailing')
axes[0].set_ylabel('Value')
axes[0].set_xticklabels(features,rotation=45,ha='right')
plt.tight_layout()


df_scaled=pd.DataFrame(scaler.fit_transform(df[features]),columns=features)
df_scaled.boxplot(ax=axes[1])
axes[1].set_title('After Min-Max-Saciling')
axes[1].set_ylabel('Value')
plt.xticks(rotation=90)
plt.tight_layout()

```

```python
audio_features=['danceability','energy','tempo','loudness','valence','acousticness','instrumentalness','speechiness','liveness']
df[audio_features]=scaler.fit_transform(df[audio_features])
```
